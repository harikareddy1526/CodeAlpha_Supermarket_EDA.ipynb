```python
product_line_counts = df["Product line"].value_counts()

product_line_counts
```

```python
plt.figure(figsize=(10, 6))

product_line_counts.plot(kind="bar")

plt.title("Number of Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Number of Transactions")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
sales_by_product = df.groupby("Product line")["Sales"].sum().sort_values(ascending=False)

sales_by_product
```

```python
plt.figure(figsize=(10, 6))

sales_by_product.plot(kind="bar")

plt.title("Total Sales by Product Line")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")

plt.tight_layout()
plt.show()
```

```python
branch_sales = df.groupby("Branch")["Sales"].sum().sort_values(ascending=False)

branch_sales
```

```python
plt.figure(figsize=(7, 5))

branch_sales.plot(kind="bar")

plt.title("Total Sales by Branch")
plt.xlabel("Branch")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
gender_sales = df.groupby("Gender")["Sales"].sum()

gender_sales
```

```python
plt.figure(figsize=(7, 5))

gender_sales.plot(kind="bar")

plt.title("Total Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
city_sales = df.groupby("City")["Sales"].sum().sort_values(ascending=False)

city_sales
```

```python
plt.figure(figsize=(7, 5))

city_sales.plot(kind="bar")

plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
customer_sales = df.groupby("Customer type")["Sales"].sum().sort_values(ascending=False)

customer_sales
```

```python
plt.figure(figsize=(7, 5))

customer_sales.plot(kind="bar")

plt.title("Total Sales by Customer Type")
plt.xlabel("Customer Type")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
payment_counts = df["Payment"].value_counts()

payment_counts
```

```python
plt.figure(figsize=(8, 5))

payment_counts.plot(kind="bar")

plt.title("Number of Transactions by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")

plt.tight_layout()
plt.show()
```

```python
df["Date"] = pd.to_datetime(df["Date"])
```

```python
df["Date"].min(), df["Date"].max()
```

```python
daily_sales = df.groupby("Date")["Sales"].sum()

daily_sales.head()
```

```python
plt.figure(figsize=(12, 6))

daily_sales.plot()

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Total Sales")

plt.tight_layout()
plt.show()
```

```python
numeric_columns = df.select_dtypes(include=np.number)

numeric_columns.corr()
```

```python
plt.figure(figsize=(10, 7))

sns.heatmap(
    numeric_columns.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Numerical Variables")

plt.tight_layout()
plt.show()
```

```python
total_sales = df["Sales"].sum()

average_sales = df["Sales"].mean()

maximum_sale = df["Sales"].max()

minimum_sale = df["Sales"].min()

total_quantity = df["Quantity"].sum()

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Maximum Sale:", maximum_sale)
print("Minimum Sale:", minimum_sale)
print("Total Quantity Sold:", total_quantity)
```

```python
top_product_line = df.groupby("Product line")["Sales"].sum().idxmax()

top_product_sales = df.groupby("Product line")["Sales"].sum().max()

print("Top Product Line:", top_product_line)
print("Sales:", top_product_sales)
```

```python
best_branch = df.groupby("Branch")["Sales"].sum().idxmax()

best_branch_sales = df.groupby("Branch")["Sales"].sum().max()

print("Best Performing Branch:", best_branch)
print("Sales:", best_branch_sales)
```
